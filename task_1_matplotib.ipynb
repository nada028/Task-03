```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.linspace(-10,10,400)
y=x**2
plt.plot(x,y)
plt.title('Quadratic Function')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
x=np.linspace(-10,10,400)
y1=np.sin(x)
y2=np.cos(x)

plt.plot(x,y1,color='red',label='sin')
plt.plot(x,y2,color='blue',label='cos')
plt.legend()

plt.title('Sin and Cos Functions')
plt.xlabel('x')
plt.ylabel('y')

plt.grid(True)

plt.show()
```

```python
integers=np.random.randint(0,100,100)
plt.scatter(np.arange(1,101),integers)
plt.title('Random Integers')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-10, 10, 400)
y1 = np.cos(x)
y2 = np.exp(-x) * np.cos(x)
y3 = 1 / (1 + np.exp(-x))
y4 = np.sin(x) / x

plt.plot(x, y1, color='red', label=r'$\cos(x)$')
plt.plot(x, y2, color='blue', label=r'$e^{-x}\cos(x)$')
plt.plot(x, y3, color='green', label=r'$1/(1+e^{-x})$')
plt.plot(x, y4, color='yellow', label=r'$\sin(x)/x$')

plt.legend()
plt.title('Various Functions')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)

plt.ylim(-2, 2)

plt.show()
```

```python
x=np.linspace(-10,10,400)
y=np.linspace(-10,10,400)
X , Y= np.meshgrid(x,y)

z=np.sin(np.sqrt(X**2+Y**2))

fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')
ax.plot_surface(X,Y,z,cmap='viridis')
plt.title('3D Surface Plot')
plt.show()
```

```python
i=np.arange(50)
j=np.arange(50)
I,J=np.meshgrid(i,j,indexing='ij')

z=np.sin(I)*np.cos(J)

plt.imshow(z,cmap='RdBu',interpolation='bicubic')
plt.colorbar()
plt.title('2D Heatmap')
plt.xlabel('J')
plt.ylabel('I')
plt.show()
```
